# Passo 3: NMF (non-Negative Matrix Factorization)

```python
import pandas as pd
import numpy as np
from sklearn.datasets import fetch_20newsgroups 
from sklearn.feature_extraction.text import TfidfVectorizer 
from sklearn.decomposition import NMF
```

```python
df = pd.read_csv("data/processed/trec_clean.csv")
df = df.dropna(subset=["clean_text"])
text_data = df["clean_text"].tolist()
display(df)
```

```python
vectorizer = TfidfVectorizer(max_features=1500, min_df=10, stop_words='english') 
X = vectorizer.fit_transform(text_data) 
words = np.array(vectorizer.get_feature_names_out()) 

print(X) 
print("X = ", words)
```

```python
# Aplicando a Fatoração de Matrizes Não Negativas 
 
nmf = NMF(n_components=10, solver="mu") 
W = nmf.fit_transform(X) 
H = nmf.components_ 

for i, topic in enumerate(H): 
     print("Tópico {}: {}".format(i + 1, ",".join([str(x) for x in words[topic.argsort()[-10:]]])))
```
